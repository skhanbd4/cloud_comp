git add module4
```

4. Commit the changes

```
git commit -m "Module 4 submission"
```

5. Push the changes

```
git push -u origin master
```

Enter your MU credentials if prompted, just an Step 2.


## Step 6: Pull down the Changes on DSA

Our final step is to update our copy of the course repository on the DSA JupyterHub.

To do this:

1. Login to the DSA JupyterHub at https://jupyterhub.dsa.missouri.edu/
2. Open a terminal
3. Navigate to course repository
4. Pull down the changes we made on Nautilus

```
git pull origin master
```


---

# Word of Caution

<font color='Red'>
We now have two copies of the course repository, one on DSA infrastructure and one on Nautilus. **NEVER** work on both at the same time, i.e., make changes to both copies. 

Any time you want to swich between them, push your changes on one copy and pull those changes on the other. 
If you make changes on both copies without committing them, you may cause [merge conflicts](https://www.atlassian.com/git/tutorials/using-branches/merge-conflicts), which can be difficult to resolve</font>
