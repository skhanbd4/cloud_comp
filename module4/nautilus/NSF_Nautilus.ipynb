# Using the NSF Nautilus HyperCluster: A Comprehensive Guide

The **NSF National Research Platform (NRP) Nautilus** is a cutting-edge distributed computing infrastructure designed specifically for scientific research and data science applications. As a federated network of Kubernetes clusters, Nautilus provides researchers with scalable, cloud-native computing resources that span multiple institutions and geographic locations.

This guide will walk you through accessing Nautilus through two primary methods: the web-based JupyterHub interface for interactive computing, and direct cluster access via command-line tools for advanced operations.

<div class="alert alert-block alert-warning">
<strong>Important Notice:</strong> Nautilus is an externally managed research infrastructure with shared resources across multiple institutions. Performance and availability can vary due to maintenance, high usage periods, or infrastructure updates. Always plan extra time for coursework and consider backup computing options for critical deadlines.
</div>

## Understanding Nautilus Architecture

Before diving into access procedures, it's helpful to understand what Nautilus is:

- **Kubernetes-Based:** Nautilus uses Kubernetes orchestration to manage containerized workloads across distributed computing nodes
- **Federated System:** Resources are contributed by multiple universities and research institutions across the United States
- **Research-Focused:** Optimized for scientific computing, machine learning, and large-scale data analysis
- **Container-Native:** Applications run in Docker containers, providing consistent environments and easy scalability

### Step 1: NRP Portal Registration
Begin by creating your NRP account:

1. Navigate to the **[NRP Portal](https://nrp.ai/)**
2. Click **"Login"** in the top-right corner
3. Use the **same authentication method** you used for JupyterHub access
4. Complete any required profile information

**What this accomplishes:** The portal creates your user identity in the Nautilus system and provides access to configuration files and documentation.

### Step 2: Request Namespace Access
Kubernetes uses **namespaces** to organize and isolate resources between different projects and users.

**Action Required:** Post your NRP username (visible in the top-right corner after login) to the **Module 4 Teams channel**. An instructor or TA will add you to the course namespace: `gp-engine-mizzou-dsa-cloud`.

**Why this matters:** Without namespace access, you won't be able to view or create resources in the cluster.

### Step 3: Command-Line Tools Installation

#### kubectl Installation
`kubectl` is the primary command-line tool for interacting with Kubernetes clusters.

**If using JupyterHub:** Skip this step - `kubectl` is pre-installed.

**For local machine setup:** Follow the [official installation guide](https://kubernetes.io/docs/tasks/tools/) for your operating system (Windows, macOS, or Linux).

#### kubelogin Plugin Installation
The `kubelogin` plugin handles authentication with the Nautilus cluster using modern OAuth2 flows.

**Installation Options:**
- **macOS:** `brew install int128/kubelogin/kubelogin`
- **Windows:** `choco install kubelogin`
- **Linux/Universal:** Use the krew plugin manager (detailed below)

#### Detailed Installation with Krew (Recommended for Linux)

**Krew** is a plugin manager for kubectl that simplifies installation of additional tools.

**1. Install Prerequisites and Krew:**
```bash
# Update system packages
sudo apt update && sudo apt install -y git curl

# Download and install Krew
(
  set -e
  cd "$(mktemp -d)"
  OS="$(uname | tr '[:upper:]' '[:lower:]')"
  ARCH="$(uname -m)"
  case $ARCH in
    x86_64) ARCH=amd64;;
    aarch64) ARCH=arm64;;
    armv7l) ARCH=arm;;
  esac
  KREW="krew-${OS}_${ARCH}"
  curl -fsSLO "https://github.com/kubernetes-sigs/krew/releases/latest/download/${KREW}.tar.gz"
  tar zxvf "${KREW}.tar.gz"
  ./"${KREW}" install krew
)

# Add Krew to your system PATH
echo 'export PATH="${KREW_ROOT:-$HOME/.krew}/bin:$PATH"' >> ~/.bashrc
source ~/.bashrc
```

**2. Install kubelogin plugin:**
```bash
kubectl krew install oidc-login
```

**3. Verify successful installation:**
```bash
kubectl oidc-login version
```
You should see version information for the kubelogin plugin.

### Step 4: Obtain Your Kubernetes Configuration

The Kubernetes configuration file contains credentials and connection information for accessing the cluster.

**Process:**
1. Return to the **[NRP documentation portal](https://nrp.ai/documentation/userdocs/start/getting-started/)**
2. Locate the section **"Cluster access via kubectl"**
3. Click **"Download Config File"**
4. Save the file (typically named `config`) to your computer

![Configuration Download](../images/config_dl.png)

**What this file contains:**
- Cluster connection endpoints
- Authentication configuration
- Your user credentials

### Step 5: Upload Configuration to JupyterHub
**Note:** Skip this step if working on a local machine.

1. Return to your JupyterHub session at https://gp-engine.nrp-nautilus.io/
2. Use the **upload button** in the file browser
3. Select and upload your downloaded config file

### Step 6: Modify Configuration for JupyterHub Environment
**Important:** This step is only necessary when using JupyterHub due to its headless (no graphical interface) environment.

JupyterHub containers run without a traditional desktop environment, so we need to modify the authentication flow:

**Process:**
1. Open the config file by double-clicking it in JupyterHub
2. Locate the section containing authentication parameters (near the end of the file)
3. Find the line with `--token-cache-storage=keyring`
4. **Replace** `keyring` with `disk`
5. **Add** these two additional parameters:
   - `--grant-type=device-code`
   - `--skip-open-browser`

**Before modification:**
![Original Config](../images/sk_config_lab1.png)

**After modification:**
![Modified Config](../images/sk_config_lab2.png)

**Why these changes are necessary:**
- `--token-cache-storage=disk`: Stores authentication tokens as files instead of using system keyring (unavailable in containers)
- `--grant-type=device-code`: Uses a device code flow suitable for headless environments
- `--skip-open-browser`: Prevents attempts to launch a web browser (impossible in containers)

### Step 7: Deploy Configuration File
Place the configuration file in the standard Kubernetes location:

```bash
# Create the .kube directory if it doesn't exist
mkdir -p ~/.kube

# Copy the config file to the standard location
cp ~/config ~/.kube/config

# Set appropriate permissions (security best practice)
chmod 600 ~/.kube/config
```

**What this accomplishes:** Kubectl will automatically find and use the configuration file in this standard location.

### Troubleshooting Common Issues

#### Authentication Problems
- **"Connection refused":** Check that kubelogin is properly installed and your config file is in `~/.kube/config`
- **"Unauthorized":** Verify you've been added to the course namespace
- **Browser authentication fails:** Ensure you're using the same identity provider as your original NRP registration

#### Configuration Issues
- **"Plugin not found":** Reinstall kubelogin using your system's package manager
- **"Invalid config":** Re-download the config file from the NRP portal and reapply modifications
- **Path problems:** Verify that `~/.kube/config` exists and has proper permissions

#### Performance Issues
- **Slow response:** Nautilus performance varies with system load; try again during off-peak hours
- **Timeouts:** Check your internet connection and consider using a different regional hub

### Best Practices for Nautilus Usage

#### Resource Management
- **Monitor usage:** Be mindful of CPU and memory consumption
- **Clean up resources:** Delete unused pods, services, and volumes
- **Use appropriate sizing:** Don't over-allocate resources you won't use

#### Security Considerations
- **Protect credentials:** Never share your config file or authentication tokens
- **Use namespaces properly:** Only access resources in your assigned namespace
- **Regular updates:** Keep kubectl and plugins updated for security patches

#### Collaboration and Etiquette
- **Shared resources:** Remember that Nautilus is a shared research infrastructure
- **Documentation:** Document your configurations and share knowledge with teammates
- **Support channels:** Use appropriate channels for help rather than individual messages
