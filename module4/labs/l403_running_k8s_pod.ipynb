## Step 2: Run the pod

Open a terminal and specify the path to the YAML file you updated in Step 1.

**Note:** You may need to update the command below with the correct path.

```BASH
cd module4/development_files/
```

Now create a pod from the terminal using following command. 

```BASH
kubectl -n gp-engine-mizzou-dsa-cloud create -f pod_python.yml
```

### Progress Check:

You should receive an output similar to:

```
pod/pod-python-sknnh created
```


## Step 3: Check pod status


Run the command below, it will show you the status of your pod

```BASH
kubectl -n gp-engine-mizzou-dsa-cloud get pods
```


## Step 4: Copy Code and Data

Once your pod's `STATUS` changes to `Running`, you will need to copy the needed files to the pod's filesystem.

**Note:** For each file referenced below, you will need to update the path to be the correct path. These files are in the [module 4 development_files](../development_files/) directory

- Make sure you update the `pod_name` to be your actual pod name.

```BASH
kubectl -n gp-engine-mizzou-dsa-cloud cp process_image.py pod_name:home/process_image.py
kubectl -n gp-engine-mizzou-dsa-cloud cp image1.npy pod_name:home/image1.npy
kubectl -n gp-engine-mizzou-dsa-cloud cp image2.npy pod_name:home/image2.npy
```


## Step 5: Accessing a Pod

Now we will get interactive access on pod using the command below, where the `/bin/bash` is our terminal shell.

- Make sure you update the `pod_name` to be your actual pod name.

```BASH
kubectl -n gp-engine-mizzou-dsa-cloud exec -it pod_name -- /bin/bash

cd home

ls -lt
```
