# Module 4 - Nautilus Cluster Computing

## Module Overview

In this module, students will gain hands-on experience with the **Nautilus National Research Platform**, a distributed Kubernetes cluster that provides high-performance computing resources for research and education. This module emphasizes practical application of Kubernetes concepts learned in Module 3 through real-world cluster deployment and workload management on a production-scale distributed computing environment.

**What makes this module unique:** Unlike previous modules that used local or simulated environments, you will work with actual research infrastructure used by universities and research institutions nationwide. You'll deploy containerized workloads, manage persistent data storage, and run deep learning models on shared computing resources—experiencing the same workflows used in professional research computing settings.

Through comprehensive setup procedures, hands-on laboratories, and practical workload deployment, you will develop proficiency in managing Kubernetes resources on a distributed computing platform that serves as a foundation for advanced cloud-native application development and research computing.

## Learning Objectives

By the end of this module, students will be able to:
- **Explain** Nautilus platform features, architecture, and complete account setup procedures including authentication and resource allocation
- **Clone** repositories and deploy compute workloads on Nautilus using industry best practices for version control integration
- **Use** kubectl to manage cluster resources in practice with real Kubernetes environments, including resource monitoring and troubleshooting
- **Configure and deploy** Kubernetes workloads including Pods, Jobs, and Persistent Volumes with appropriate resource specifications
- **Run** analysis pipelines on Kubernetes and Nautilus clusters with proper resource management and cleanup procedures
- **Implement** container-based data processing workflows with Python and scientific computing libraries (NumPy, PyTorch)
- **Demonstrate** proficiency through practical hands-on labs, deep learning model training, and resource optimization strategies

## Module Topics

### 1. Nautilus Platform Architecture and National Research Platform
* **Platform Overview**: Nautilus as part of the National Research Platform (NRP) ecosystem—a federated network of computing resources spanning multiple universities and research institutions
* **Distributed Computing**: Multi-institutional cluster federation, resource sharing across geographic locations, and how computational workloads are distributed
* **Research Applications**: Scientific computing, large-scale data analysis, machine learning model training, and collaborative research workloads
* **Platform Evolution**: Historical context from Pacific Research Platform (PRP) to current Nautilus infrastructure and future development directions

### 2. Account Management and Access Control
* **Account Creation**: Registration process through institutional authentication, project allocation workflows, and approval procedures
* **Access Methods**: Web-based interface (JupyterHub), kubectl command-line access from local machines, and programmatic API interactions
* **Security Model**: Role-based access control (RBAC) for permission management, namespace isolation for multi-tenant security, and resource quotas to ensure fair usage
* **Authentication**: Token management for secure access, kubectl configuration files, cluster contexts, and maintaining secure cluster connections

### 3. Repository Integration and Development Workflow
* **Version Control Integration**: Git repository cloning to cluster storage, synchronization between local and remote repositories, and collaborative development patterns
* **Development Environment**: Jupyter notebook deployment on Nautilus, interactive code development workflows, and notebook-based experimentation
* **Code Management**: SSH key configuration for seamless authentication, HTTPS credential alternatives, and repository access patterns for secure code deployment
* **File System Management**: Persistent volume claims (PVCs) for data persistence, data staging workflows, artifact storage strategies, and managing data across pod lifecycles

### 4. Kubernetes Workload Deployment and Management
* **Pod Deployment**: Container specification using YAML manifests, resource allocation (CPU/memory requests and limits), and pod lifecycle management (creation, monitoring, termination)
* **YAML Configuration**: Writing and understanding Pod manifests, configuring environment variables for application settings, and volume mounting for data access
* **Resource Management**: Setting appropriate CPU and memory allocation, understanding limits vs. requests, quality of service (QoS) classes, and avoiding resource conflicts
* **Monitoring and Troubleshooting**: Log analysis using `kubectl logs`, pod status monitoring with `kubectl get` and `kubectl describe`, and systematic debugging techniques for failed deployments

### 5. Scientific Computing and Data Processing Workflows
* **Container-based Analytics**: Deploying Python scientific computing stack (NumPy, SciPy, Matplotlib, PyTorch) within containers for reproducible analysis
* **Data Pipeline Implementation**: Handling input data from persistent volumes, performing data transformations and analysis, and writing results back to storage
* **Image Processing and Deep Learning Applications**: NumPy array processing for scientific data, training convolutional neural networks (CNNs) for image classification, and evaluating model performance
* **Performance Optimization**: Monitoring resource utilization during computation, implementing parallel processing where appropriate, and understanding scalability considerations for larger workloads

## Important Usage Guidelines

### Nautilus Platform Policies
- **Shared Infrastructure**: Nautilus is a shared research platform used by students and researchers nationwide—be considerate of other users
- **Resource Termination**: **CRITICAL** - All pods, jobs, and persistent volumes MUST be terminated after use to free resources for others
- **Academic Use Only**: Platform access is restricted to course-related activities and educational research—no personal or commercial use
- **Data Management**: Follow appropriate data handling practices for scientific datasets, respecting licensing and usage restrictions

### Cluster Best Practices
- **Resource Optimization**: Use appropriate resource requests and limits for containers—avoid over-requesting resources you won't use
- **Monitoring**: Regularly check pod status and resource utilization during execution to identify issues early
- **Troubleshooting**: Use `kubectl logs <pod-name>` and `kubectl describe pod <pod-name>` commands for systematic debugging
- **Community Guidelines**: Follow Nautilus community standards for shared resource usage, including cleanup and respectful communication

### Common Pitfalls to Avoid
- **Forgetting Cleanup**: Leaving pods or jobs running after lab completion wastes resources and may impact your quota
- **Last-Minute Work**: Starting assignments close to deadlines without accounting for approval times or cluster load
- **Ignoring Errors**: Skipping error messages without investigating—errors usually provide specific clues for resolution
- **Excessive Resource Requests**: Requesting more CPU/memory than needed can delay pod scheduling
- **Not Backing Up Work**: Forgetting that PVC data is temporary and will be deleted after course completion
