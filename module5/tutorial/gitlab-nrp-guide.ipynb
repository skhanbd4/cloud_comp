> **Complete Guide for Working with NRP GitLab using GP-Engine Terminal**  
> This tutorial walks you through initializing a new repository, generating and configuring SSH keys (with no existing keys), and pushing your first file using https://gitlab.nrp-nautilus.io/ from the GP-Engine terminal.

---

## Overview

This guide provides step-by-step instructions for:
1. Registering and creating a GitLab repository on NRP Nautilus
2. Generating a new SSH key pair (from scratch)
3. Adding your public key to your NRP GitLab account
4. Configuring your Git identity
5. Working with empty repositories using GitLab's recommended workflows
6. Staging, committing, and pushing your first file

We assume you are working in a fresh GP-Engine Nautilus terminal, with no existing SSH key pairs.

---

## Prerequisites

- Active account on https://gitlab.nrp-nautilus.io/
- Access to a terminal on https://gp-engine.nrp-nautilus.io/
- Some command-line basics
- Familiarity with Git fundamentals

## Step 1: Register and Create a New Project on NRP GitLab

1. Go to https://gitlab.nrp-nautilus.io/ and register/login.
2. Click **"New Project"** or **"Create blank project"**.
3. Fill in project details:
   - **Project name:** `cloud-computing-test` (or your choice)
   - **Visibility Level:** Public or Private
   - Leave other options default
4. Click **"Create project"**.

After creating your project, you'll see a page that says **"The repository for this project is empty"** with several command-line options to get started. We'll follow these instructions in the steps below.

5. On your new project page, locate the SSH URL. It will look something like:
   ```
   ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/cloud-computing-test.git
   ```

## Step 2: Generate New SSH Key Pair (No Existing Key)

### 2.1 Check for Existing Keys

In your GP-Engine terminal, check for existing keys:
```bash
ls -la ~/.ssh
```

> **Note:** You may see one of the following:
> - A list of existing SSH key files (like `id_ed25519`, `id_rsa`)
> - An error message: `ls: cannot access '/home/jovyan/.ssh': No such file or directory`
> 
> If you see the error or no key files, proceed to generate a new key.

### 2.2 Generate a New SSH Key

Generate a new Ed25519 key (recommended for security):
```bash
ssh-keygen -t ed25519 -C "your.email@umsystem.edu"
```

**Replace** `your.email@umsystem.edu` with your actual email address.

**At the prompts:**
- **"Enter file in which to save the key"**: Press **Enter** to accept the default location
- **"Enter passphrase"**: Press **Enter** for no passphrase, or enter a secure passphrase
- **"Enter same passphrase again"**: Confirm by pressing **Enter** again or re-entering your passphrase

### 2.3 Start SSH Agent and Add Key

```bash
eval "$(ssh-agent -s)"
ssh-add ~/.ssh/id_ed25519
```

**Expected output:**
```
Agent pid 12345
Identity added: /home/jovyan/.ssh/id_ed25519 (your.email@umsystem.edu)
```

### 2.4 Display Your Public Key

```bash
cat ~/.ssh/id_ed25519.pub
```

**Copy the entire output** (single long line starting with `ssh-ed25519`). You'll need this for the next step.

---

## Step 3: Add SSH Public Key to NRP GitLab

1. In your browser, go to https://gitlab.nrp-nautilus.io/
2. Click your **avatar/photo** (upper right) → **Preferences** or **Edit Profile**
3. In the left menu, select **SSH Keys**
4. In the **"Key"** field, paste your entire public key (from Step 2.4)
5. Add a **Title** (e.g., "GP-Engine Terminal Key")
6. Click **"Add key"**

For more information, see [How to use SSH keys?](https://docs.gitlab.com/ee/user/ssh.html) in GitLab's documentation.

## Step 4: Configure Git Identity

GitLab provides options to configure your Git identity either **locally** (for this project only) or **globally** (for all projects). For this tutorial, we'll use **local** configuration.

### Local Git Configuration

Set your name and email for this specific project:

```bash
git config --local user.name "Your Full Name"
git config --local user.email "your.email@umsystem.edu"
```

**Replace** with your actual name and email address.

> **Tip:** If you want to configure Git globally (for all repositories on your system), use:
> ```bash
> git config --global user.name "Your Full Name"
> git config --global user.email "your.email@umsystem.edu"
> ```

## Step 5: Working with Your Empty Repository

GitLab provides three main workflows when starting with an empty repository. Choose the one that fits your situation:

### Option 1: Create a New Repository (Recommended for Beginners)

This is the recommended approach when starting from scratch.

```bash
# Clone the empty repository
git clone ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/cloud-computing-test.git
cd cloud-computing-test

# Create and switch to the main branch
git switch --create main

# Create a README file
touch README.md

# Add content to README
echo "# Cloud Computing Test" > README.md
echo "" >> README.md
echo "This repository is for practice with GitLab and SSH on Nautilus NRP." >> README.md

# Stage and commit the file
git add README.md
git commit -m "add README"

# Push to GitLab
git push --set-upstream origin main
```

**Replace** `username` with your actual GitLab username in the clone URL.

---

### Option 2: Push an Existing Folder

If you already have a local folder with files you want to version control:

```bash
# Navigate to your existing folder
cd existing_folder

# Initialize Git repository with main as the default branch
git init --initial-branch=main

# Add the remote repository
git remote add origin ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/cloud-computing-test.git

# Configure Git identity locally (if not done globally)
git config --local user.name "Your Full Name"
git config --local user.email "your.email@umsystem.edu"

# Stage all files
git add .

# Commit the files
git commit -m "Initial commit"

# Push to GitLab
git push --set-upstream origin main
```

**Replace** `username` with your actual GitLab username.

---

### Option 3: Push an Existing Git Repository

If you have an existing Git repository from another platform (e.g., GitHub) or location:

```bash
# Navigate to your existing repository
cd existing_repo

# Rename the current remote (optional, preserves old remote)
git remote rename origin old-origin

# Add the new GitLab remote
git remote add origin ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/cloud-computing-test.git

# Push all branches
git push --set-upstream origin --all

# Push all tags
git push --set-upstream origin --tags
```

**Replace** `username` with your actual GitLab username.

After pushing your changes:

1. Go to https://gitlab.nrp-nautilus.io/
2. Navigate to **Your project** → **Repository** → **Files**
3. You should see:
   - Your `README.md` file (or other files you pushed)

## Understanding the SSH URL Format

The NRP GitLab SSH URL has a specific format:
```
ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/repository-name.git
```

**Components:**
- `ssh://` - Protocol (SSH)
- `git@gitlab-ssh.nrp-nautilus.io` - GitLab server
- `userid` - SSH port (different from standard port 22)
- `username` - Your GitLab username
- `repository-name.git` - Your repository name

> **Important:** Always use the SSH URL provided by GitLab on your project page to ensure correct configuration.

## Troubleshooting

### Permission Denied (publickey)

**Problem:** `Permission denied (publickey)` when cloning or pushing

**Solution:**
1. Verify SSH key is added to GitLab (Settings → SSH Keys)
2. Ensure SSH agent is running and key is loaded:
   ```bash
   eval "$(ssh-agent -s)"
   ssh-add ~/.ssh/id_ed25519
   ```
3. Test SSH connection:
   ```bash
   ssh -T git@gitlab-ssh.nrp-nautilus.io -p userid
   ```

### Cannot Connect to Repository

**Problem:** `Could not resolve hostname` or connection timeout

**Solution:**
- Verify you're using the correct SSH URL from your GitLab project page
- Check that you're connected to the network
- Ensure the port number (userid) is included in the URL

### Wrong Branch Name

**Problem:** Local branch is different from what GitLab expects

**Solution:**
```bash
# Rename current branch to main
git branch -M main

# Push to main
git push --set-upstream origin main
```

### Remote Already Exists

**Problem:** `fatal: remote origin already exists`

**Solution:**
```bash
# Remove existing remote
git remote remove origin

# Add the correct remote
git remote add origin ssh://git@gitlab-ssh.nrp-nautilus.io:userid/username/cloud-computing-test.git
```

### SSH Agent Not Running

**Problem:** `Could not open a connection to your authentication agent`

**Solution:**
```bash
# Start the SSH agent
eval "$(ssh-agent -s)"

# Add your key
ssh-add ~/.ssh/id_ed25519
```

## Best Practices

1. **Commit Often:** Make small, focused commits with clear messages
2. **Pull Before Push:** Always pull the latest changes before pushing to avoid conflicts
3. **Use Meaningful Commit Messages:** Describe what and why, not just what
4. **Protect Your SSH Keys:** Never share your private key (`id_ed25519`)
5. **Test SSH Connection:** Verify SSH setup before starting work
6. **Branch for Features:** Use branches for new features or experiments
7. **Review Before Commit:** Use `git status` and `git diff` to review changes
